Bienvenido a la sesión práctica manual. Al igual que hicimos con las Convoluciones, vamos a abrir la **"caja negra"** de las Redes Neuronales Recurrentes (RNN).

El objetivo de hoy no es usar librerías avanzadas como TensorFlow o PyTorch todavía, sino entender **matemáticamente e intuitivamente** cómo una red puede recordar información del pasado para predecir el futuro.



## 1. El Problema: Datos Secuenciales y Memoria

Imaginemos que estamos monitoreando la temperatura de una máquina cada hora:

**Secuencia:** `[1, 2, 3]`

Nuestro objetivo es predecir el siguiente valor.

Hasta aquí parece un problema clásico de regresión… pero hay un detalle clave:

👉 **El orden importa.**

No es lo mismo ver:

- `[1, 2, 3]`
- `[3, 2, 1]`

En series de tiempo:

- El pasado influye en el presente.
- Las tendencias se acumulan.
- Los patrones dependen del contexto previo.

### Ejemplos reales en BI

- Ventas diarias  
- Tráfico web  
- Demanda energética  
- GMV semanal  
- Open Rate de campañas  

Todos tienen **dependencia temporal**.



## ¿Por qué no usar una Red Neuronal Tradicional?

Supongamos que usamos una red Feedforward.

Le damos como entrada el valor `3`.

La red intenta predecir el siguiente valor… pero hay un problema:

Para la red, el `3` es un dato aislado.

No sabe que antes hubo:

`1 → luego 2 → luego 3`

Es como preguntarle a alguien:

> “¿Qué viene después de 3?”  
> … sin decirle que estamos contando.

La red no tiene contexto.  
No tiene memoria.  
No tiene historia.

📌 **Las redes tradicionales modelan relaciones, pero no secuencias.**



## 2. La Arquitectura y las Fórmulas

Una RNN procesa los datos paso a paso (`t`). En cada paso:

1️⃣ Mira el dato actual (`x_t`)  
2️⃣ Recuerda lo que pasó antes (`h_{t-1}`)

### Ecuaciones del “Cerebro” Recurrente

$$
h_t = \tanh(W_{xh} x_t + W_{hh} h_{t-1} + b_h)
$$

$$
y_t = W_{hy} h_t + b_y
$$

### Intuición del Estado Oculto

El estado oculto es como un **resumen ejecutivo del pasado**.

No guarda cada dato histórico…  
Guarda una representación comprimida de lo relevante.

Podemos pensarlo como:

> **h_t = “todo lo que la red considera importante hasta el tiempo t”**



## 3. Simulación “A Mano”

Pesos definidos:

- W_xh = 0.5  
- W_hh = 0.8  
- W_hy = 1.0  
- h_0 = 0  


```python

import numpy as np

# Input sequence
inputs = [1, 2, 3]

# Fixed weights
W_xh = 0.5
W_hh = 0.8
W_hy = 1.0

# Initial hidden state
h_t = 0

print(f"Initial state h_0: {h_t}")
print("-" * 30)

hidden_states = []

# Forward pass through time
for i, x in enumerate(inputs):
    t = i + 1
    
    print(f"\n[Time t={t}] Input x: {x}, Previous memory h_{t-1}: {h_t:.3f}")
    
    # Core RNN equation
    h_t = np.tanh(W_xh * x + W_hh * h_t)
    
    hidden_states.append(h_t)
    print(f"   -> New hidden state h_{t}: {h_t:.3f}")

# Final prediction
y_pred = W_hy * h_t

print("-" * 30)
print(f"\nFinal prediction: {y_pred:.3f}")
```


## 4. Interpretación

Observación clave:

La red **no** predice el siguiente valor real de la secuencia.

¿Por qué?

Porque los pesos no han sido entrenados.

La red tiene memoria…  
pero aún no sabe cómo usarla para predecir correctamente.

Aquí entra el entrenamiento:

**Backpropagation Through Time (BPTT)**
